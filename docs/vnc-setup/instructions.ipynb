# VNC Reproducibility Kit Setup

This "hacks" kit is designed to bridge the gap between a macOS workstation and a headless Ubuntu server (MACS-HPC) to create a unified development environment. It automates the complex configuration of a persistent, high-performance VNC session that mirrors local UI preferences while ensuring remote server resources (like GNOME sessions and D-Bus) remain active even after logout.

## Core Objectives
* Persistent Sessions: Configures systemd lingering so the VNC server and your applications (like VS Code or simulation tools) keep running even when you aren't connected.

* Visual Consistency: Synchronizes the "Yaru" appearance and icon sets across platforms so the remote desktop feels like a local extension of the Mac.

* Automated Security: Handles the SSH key-pair handshake and configures a secure tunnel, eliminating the need to manually manage ports or passwords.

* Environment Recovery: Provides a single-command "clean start" script to resolve common GNOME crashes or lock-screen freezes without requiring a full server reboot.

## Content

| File | Purpose |
| :--- | :--- |
| `server-setup.sh` | Run on the server — installs vnc-start-clean, xstartup, systemd service, linger, and Yaru appearance |
| `mac-setup.sh` | Run on a fresh Mac — generates SSH key, copies it to server, writes SSH config and connect script |
| `connect-vnc.zsh` | Snapshot of the working connect script |

## To rebuild everything from scratch: 

assume `vnc-setup` folder is in home

### On Mac:

* `~/vnc-setup/mac-setup.sh`

### On Server (after SSH key is in place)

* `scp ~/vnc-setup/server-setup.sh ubuntu-server:~/`
* `ssh ubuntu-server "bash ~/server-setup.sh"`

**Note:** The session memory is saved—future contexts will retain these setup details.
